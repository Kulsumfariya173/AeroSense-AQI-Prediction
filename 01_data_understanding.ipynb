```python
feature_cols = [
    "City", "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene", "Toluene",
    "Year", "Month", "Season"
]

X_train = train_df[feature_cols].copy()
y_train = train_df["AQI"].copy()

X_test = test_df[feature_cols].copy()
y_test = test_df["AQI"].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
print("Number of cities:", X_train["City"].nunique())
print("Cities:", X_train["City"].unique())

print("\nSeasons:", X_train["Season"].unique())
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(X_train[["City", "Season"]])

print("Encoder fitted successfully.")
```

```python
X_train_cat = encoder.transform(X_train[["City", "Season"]])
X_test_cat = encoder.transform(X_test[["City", "Season"]])

print("Encoded training shape:", X_train_cat.shape)
print("Encoded testing shape:", X_test_cat.shape)
```

```python
encoded_feature_names = encoder.get_feature_names_out(["City", "Season"])

print(encoded_feature_names)
```

```python
numeric_cols = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene", "Toluene",
    "Year", "Month"
]

X_train_num = X_train[numeric_cols].copy()
X_test_num = X_test[numeric_cols].copy()

print("Numerical training shape:", X_train_num.shape)
print("Numerical testing shape:", X_test_num.shape)
```

```python
X_train_num.describe().T[["min", "max"]]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(X_train_num)

print("Scaler fitted successfully.")
```

```python
X_train_num_scaled = scaler.transform(X_train_num)
X_test_num_scaled = scaler.transform(X_test_num)

print("Scaled training shape:", X_train_num_scaled.shape)
print("Scaled testing shape:", X_test_num_scaled.shape)
```

```python
import numpy as np

X_train_final = np.hstack([X_train_num_scaled, X_train_cat])
X_test_final = np.hstack([X_test_num_scaled, X_test_cat])

print("Final training shape:", X_train_final.shape)
print("Final testing shape:", X_test_final.shape)
```

```python
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("Missing y_train:", y_train.isnull().sum())
print("Missing y_test:", y_test.isnull().sum())
```

```python
cluster_cols = [
    "PM2.5", "PM10", "NO", "NO2", "NOx",
    "NH3", "CO", "SO2", "O3", "Benzene", "Toluene"
]

X_cluster = model_df[cluster_cols].copy()

print("Clustering dataset shape:", X_cluster.shape)
print("Missing values:")
print(X_cluster.isnull().sum())
```

```python
cluster_medians = X_cluster.median()

X_cluster_imputed = X_cluster.fillna(cluster_medians)

print("Missing values after imputation:")
print(X_cluster_imputed.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

cluster_scaler = StandardScaler()

X_cluster_scaled = cluster_scaler.fit_transform(X_cluster_imputed)

print("Scaled clustering shape:", X_cluster_scaled.shape)
print("Mean of scaled features:", X_cluster_scaled.mean(axis=0))
print("Standard deviation:", X_cluster_scaled.std(axis=0))
```

```python
from sklearn.cluster import KMeans

inertia_values = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_cluster_scaled)
    inertia_values.append(kmeans.inertia_)

print("K values:", list(range(2, 11)))
print("Inertia values:")
print(inertia_values)
```

```python
from sklearn.metrics import silhouette_score

silhouette_values = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    cluster_labels = kmeans.fit_predict(X_cluster_scaled)
    score = silhouette_score(X_cluster_scaled, cluster_labels)
    silhouette_values.append(score)

print("K values:", list(range(2, 11)))
print("Silhouette scores:")
print(silhouette_values)
```

```python
kmeans_final = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans_final.fit_predict(X_cluster_scaled)

print("K-Means clustering completed!")
print("Number of observations:", len(cluster_labels))
print("Number of clusters:", len(set(cluster_labels)))
```

```python
import pandas as pd

cluster_counts = pd.Series(cluster_labels).value_counts().sort_index()

print("Observations per cluster:")
print(cluster_counts)
```

```python
cluster_profile = X_cluster_imputed.copy()
cluster_profile["Cluster"] = cluster_labels

cluster_means = cluster_profile.groupby("Cluster").mean()

print("Average pollutant values by cluster:")
print(cluster_means)
```

```python
cluster_aqi = model_df[["AQI"]].copy()
cluster_aqi["Cluster"] = cluster_labels

print("Average AQI by cluster:")
print(cluster_aqi.groupby("Cluster")["AQI"].mean())

print("\nMedian AQI by cluster:")
print(cluster_aqi.groupby("Cluster")["AQI"].median())
```

```python
# Get the original row indices from our train/test split
train_indices = train_df.index
test_indices = test_df.index

# Get pollutant data for train and test
X_cluster_train = model_df.loc[train_indices, cluster_cols].copy()
X_cluster_test = model_df.loc[test_indices, cluster_cols].copy()

print("Training clustering data:", X_cluster_train.shape)
print("Testing clustering data:", X_cluster_test.shape)
```

```python
# Calculate medians using TRAINING data only
cluster_train_medians = X_cluster_train.median()

# Impute both train and test using training medians
X_cluster_train_imputed = X_cluster_train.fillna(cluster_train_medians)
X_cluster_test_imputed = X_cluster_test.fillna(cluster_train_medians)

print("Training missing values:")
print(X_cluster_train_imputed.isnull().sum())

print("\nTesting missing values:")
print(X_cluster_test_imputed.isnull().sum())
```

```python
cluster_scaler_proper = StandardScaler()

X_cluster_train_scaled = cluster_scaler_proper.fit_transform(
    X_cluster_train_imputed
)

X_cluster_test_scaled = cluster_scaler_proper.transform(
    X_cluster_test_imputed
)

print("Training scaled shape:", X_cluster_train_scaled.shape)
print("Testing scaled shape:", X_cluster_test_scaled.shape)

print("\nTraining means:")
print(X_cluster_train_scaled.mean(axis=0))

print("\nTraining standard deviations:")
print(X_cluster_train_scaled.std(axis=0))
```

```python
kmeans_proper = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans_proper.fit(X_cluster_train_scaled)

print("Proper K-Means trained successfully!")
print("Cluster centers shape:", kmeans_proper.cluster_centers_.shape)
```

```python
cluster_train_labels = kmeans_proper.predict(X_cluster_train_scaled)
cluster_test_labels = kmeans_proper.predict(X_cluster_test_scaled)

print("Training cluster labels:", cluster_train_labels.shape)
print("Testing cluster labels:", cluster_test_labels.shape)

print("\nTraining cluster counts:")
print(pd.Series(cluster_train_labels).value_counts().sort_index())

print("\nTesting cluster counts:")
print(pd.Series(cluster_test_labels).value_counts().sort_index())
```

```python
cluster_train_labels = kmeans_proper.predict(X_cluster_train_scaled)
cluster_test_labels = kmeans_proper.predict(X_cluster_test_scaled)

print("Training cluster labels:", cluster_train_labels.shape)
print("Testing cluster labels:", cluster_test_labels.shape)

print("\nTraining cluster counts:")
print(pd.Series(cluster_train_labels).value_counts().sort_index())

print("\nTesting cluster counts:")
print(pd.Series(cluster_test_labels).value_counts().sort_index())
```

```python
# Create an empty array for predictions
y_pred_rf_cluster_specific = np.zeros(len(X_test_clustered))

# Find test observations belonging to each cluster
test_cluster_0_idx = (cluster_test_labels == 0)
test_cluster_1_idx = (cluster_test_labels == 1)

# Predict using the corresponding cluster-specific model
y_pred_rf_cluster_specific[test_cluster_0_idx] = rf_model_cluster0.predict(
    X_test_final[test_cluster_0_idx]
)

y_pred_rf_cluster_specific[test_cluster_1_idx] = rf_model_cluster1.predict(
    X_test_final[test_cluster_1_idx]
)

print("Cluster-specific predictions completed!")
print("Total predictions:", len(y_pred_rf_cluster_specific))
```

```python
def get_aqi_category(aqi):
    if aqi <= 50:
        return "Good"
    elif aqi <= 100:
        return "Satisfactory"
    elif aqi <= 200:
        return "Moderate"
    elif aqi <= 300:
        return "Poor"
    elif aqi <= 400:
        return "Very Poor"
    else:
        return "Severe"


# Test the function
test_values = [25, 75, 150, 250, 350, 450, 1000]

for value in test_values:
    print(value, "→", get_aqi_category(value))
```

```python
health_alerts = {
    "Good": {
        "message": "Air quality is considered good. Air pollution poses little or no risk.",
        "advice": "No special precautions are generally needed."
    },

    "Satisfactory": {
        "message": "Air quality is acceptable, but some pollutants may affect sensitive individuals.",
        "advice": "Sensitive people should consider reducing prolonged outdoor exposure if they experience discomfort."
    },

    "Moderate": {
        "message": "Air quality may cause discomfort to sensitive individuals.",
        "advice": "People with respiratory or heart conditions should consider limiting prolonged outdoor exertion."
    },

    "Poor": {
        "message": "Poor air quality may affect health, especially for sensitive groups.",
        "advice": "Sensitive individuals should reduce prolonged or strenuous outdoor activity."
    },

    "Very Poor": {
        "message": "Very poor air quality can cause respiratory health effects.",
        "advice": "Avoid prolonged or strenuous outdoor activity, especially if you are sensitive to air pollution."
    },

    "Severe": {
        "message": "Severe air pollution can affect health and may cause serious respiratory effects.",
        "advice": "Avoid unnecessary outdoor exposure and consider appropriate protective measures."
    }
}

print("Health alert messages created successfully!")

for category, alert in health_alerts.items():
    print(f"\n{category}:")
    print("Message:", alert["message"])
    print("Advice:", alert["advice"])
```

```python
def get_health_alert(predicted_aqi):
    category = get_aqi_category(predicted_aqi)
    
    alert = health_alerts[category]
    
    return {
        "AQI": round(predicted_aqi, 2),
        "Category": category,
        "Message": alert["message"],
        "Advice": alert["advice"]
    }


# Test with one example
result = get_health_alert(275)

print(result)
```

```python
#Test the alert system across all AQI categories
 test_aqi_values = [30, 75, 150, 250, 350, 450]

for aqi in test_aqi_values:
    result = get_health_alert(aqi)

    print(
        f"AQI: {result['AQI']} | "
        f"Category: {result['Category']}"
    )
```
